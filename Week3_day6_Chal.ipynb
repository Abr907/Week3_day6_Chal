```python
#Part 1 — Create the Text class
class Text:
    """Analyze and manage text."""

    def __init__(self, text):
        """Initialize the object with a string."""
        if not isinstance(text, str):
            raise TypeError("text must be a string")
        self.text = text

    def word_frequency(self, word):
        """Return the case-insensitive count of a word, or None if absent."""
        words = self.text.lower().split()
        target = word.lower()

        count = words.count(target)
        return count if count > 0 else None

    def most_common_word(self):
        """Return the most frequent word, or None if the text is empty."""
        words = self.text.lower().split()

        if not words:
            return None

        frequencies = {}
        for word in words:
            frequencies[word] = frequencies.get(word, 0) + 1

        return max(frequencies, key=frequencies.get)

    def unique_words(self):
        """Return a list of unique words, ignoring case."""
        words = self.text.lower().split()
        return list(dict.fromkeys(words))

    @classmethod
    def from_file(cls, file_path):
        """Create an instance from a file, or return None if not found."""
        try:
            with open(file_path, "r", encoding="utf-8") as file:
                content = file.read()
            return cls(content)
        except FileNotFoundError:
            print(f"Error: File not found: {file_path}")
            return None
```

```python
# Part 2 — Create the TextModification class

import string
import re


class TextModification(Text):
    """Extend Text with text-cleaning methods."""

    STOP_WORDS = {
        "a", "an", "the", "and", "or", "but", "is", "are",
        "was", "were", "be", "been", "being", "in", "on",
        "at", "to", "of", "for", "with", "by", "as", "it",
        "this", "that", "these", "those", "i", "you", "he",
        "she", "we", "they", "my", "your", "our", "their",
        "am", "do", "does", "did", "have", "has", "had",
        "from", "not", "so", "if", "than", "then", "too"
    }

    def remove_punctuation(self):
        """Remove punctuation characters and return the cleaned text."""
        translator = str.maketrans("", "", string.punctuation)
        self.text = self.text.translate(translator)
        return self.text

    def remove_stop_words(self):
        """Remove common English stop words and return the cleaned text."""
        words = self.text.split()
        filtered_words = [
            word for word in words
            if word.lower() not in self.STOP_WORDS
        ]
        self.text = " ".join(filtered_words)
        return self.text

    def remove_special_characters(self):
        """Remove characters other than letters, digits, and whitespace."""
        self.text = re.sub(r"[^a-zA-Z0-9\s]", "", self.text)
        return self.text
```

```python
# Part 3 — Test the Text class

sample_text = "Python is great. Python is easy to learn. Data is useful!"

my_text = Text(sample_text)

print("Original text:", my_text.text)
print("Frequency of 'Python':", my_text.word_frequency("Python"))
print("Frequency of 'Java':", my_text.word_frequency("Java"))
print("Most common word:", my_text.most_common_word())
print("Unique words:", my_text.unique_words())
```
